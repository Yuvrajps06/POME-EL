# PricePilot: Exploratory Data Analysis & Market Intelligence

This notebook exercises the reusable EDA plotting routines from  using the processed M5 dataset.


```python
import sys
from pathlib import Path
import pandas as pd
sys.path.append("..")

from src.data.eda import (
    compute_market_kpis,
    plot_sales_and_price_over_time,
    plot_price_vs_quantity,
    plot_sales_distribution,
    plot_seasonality,
    plot_price_change_vs_sales_change,
    plot_promotion_effect,
    plot_product_comparison,
)
from src.utils.config import DATA_PROCESSED_DIR, PRIMARY_MODELING_ITEM

# Load processed dataset
data_path = DATA_PROCESSED_DIR / "m5_processed.parquet"
if not data_path.exists():
    data_path = DATA_PROCESSED_DIR / "m5_processed.csv"
df = pd.read_parquet(data_path) if str(data_path).endswith(".parquet") else pd.read_csv(data_path)
df["date"] = pd.to_datetime(df["date"])
print(f"Loaded {len(df):,} records for {df['item_id'].nunique()} items and {df['store_id'].nunique()} stores.")
```

## 1. Market Intelligence KPIs

```python
kpis = compute_market_kpis(df)
kpis.head(15)
```

## 2. Sales and Price Dynamics Over Time

```python
fig_ts = plot_sales_and_price_over_time(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1")
fig_ts.show()
```

## 3. Empirical Demand Curve (Price vs Quantity)

```python
fig_demand_nat = plot_price_vs_quantity(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1", log_scale=False)
fig_demand_nat.show()

fig_demand_log = plot_price_vs_quantity(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1", log_scale=True)
fig_demand_log.show()
```

## 4. Sales Distribution

```python
fig_dist = plot_sales_distribution(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1")
fig_dist.show()
```

## 5. Seasonality Analysis (Day of Week, Month, Events)

```python
fig_dow = plot_seasonality(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1", freq="dow")
fig_dow.show()

fig_month = plot_seasonality(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1", freq="month")
fig_month.show()

fig_event = plot_seasonality(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1", freq="event")
fig_event.show()
```

## 6. Price Revision Response & Promotional Uplift

```python
fig_pct = plot_price_change_vs_sales_change(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1")
fig_pct.show()

fig_promo = plot_promotion_effect(df, item_id=PRIMARY_MODELING_ITEM, store_id="CA_1")
fig_promo.show()
```

## 7. Product Comparisons Across Stores

```python
fig_comp_sales = plot_product_comparison(df, metric="sales")
fig_comp_sales.show()

fig_comp_price = plot_product_comparison(df, metric="price")
fig_comp_price.show()
```
